# Ex 2 Primality Test 

```python
from math import sqrt
def is_prime(n: int) -> bool:
    if abs(n)==1:
        return False
    for i in range(2,int(sqrt(n))):
        if n%i:
            return False
    return True
```

```python
for x in range(1,30):
    if is_prime(x):
        print(x,end=" ,")
```

# Ex 3 Modular Arithmetic 

```python
def mod_ops(a, b, m) -> dict :
    D={
        "add":(a+b)%m,
        "sub":(a-b)%m,
        "mul":(a*b)%m
    }
    return D
```

```python
a=17
b=9 
m=5
D=mod_ops(a,b,m)
```

# Ex 4 GCD and LCM 

```python
def lcm(a, b):
    ma=a
    mb=b
    while ma!=mb:
        if ma<mb:
            ma+=a
        else:
            mb+=b
    return ma
```

```python
Tests=[(12,8),(100,75),(17,13)]
```

```python
[lcm(a,b) for a,b in Tests]
```

```python
from math import gcd
```

```python
all([gcd(a,b)*lcm(a,b)==a*b for a,b in Tests])
```

# Ex 5 Fibonacci Sequence 

```python
def  fibonacci(n: int) -> list[int]:
    if n<=0:
        return []
    if n==1:
        return [0]
    F=[0,1]
    for i in range(2,n):
        F.append(F[i-1]+F[i-2])
    return F
```

```python
print(fibonacci(2) )
```

# Ex 6 Counting Divisors 

```python
def divisors(n: int) -> list[int]:
    D=[]
    for x in range(2,n):
        if n%x==0:
            D.append(x)
    return D
```

```python
for n in [12,16,17,36]:
    print("n=" ,n,"n divisors ", divisors(n) )
```

```python
from math import log2,ceil
def power_set_v2(s: list) -> list:
    P=[set()]
    for n in range(1,2**len(s)):
        L=set()
        for i in range(ceil(log2(n))+1):
            if (n>>i)&1:
                L.add(s[i])
        P.append(L)
    return P
```

```python
power_set_v2([1,2,3])
```

```python
def power_set_rec(s: list) -> list:
    P=[set()]
    if len(s)!=0:
        Pi= power_set_rec(s[1:])
        for st in Pi:
            if st not in P:
                P.append(st)
            P.append({s[0]}.union(st))
    return P
```

```python
power_set_rec([])
```

```python
def sieve(N: int) -> list[int]:
    Prime=[False,False]+[True for i in range(N-2)]
    PrimeNumbers=[]
    p=2
    count=0
    while p*p<=N:
        if Prime[p]:
            count+=1
            for i in range(p*p,N,p):
                Prime[i]=False
        p=p+1
    for i in range(len(Prime)):
        if Prime[i]:
            PrimeNumbers.append(i)
    return PrimeNumbers
```

```python
sieve(20)
```
