# Ex 1 Basic Propositions SymPy

```python
from sympy import symbols,sqrt,pretty
```

```python
p,q=symbols("p q")
```

```python
F=p&(q|~p)
```

```python
print(pretty(p),"\t\t",pretty(q),"\t\t",pretty(F))
for vp in [False,True]:
    for vq in [False,True]:
        print("p=",vp,"\t","q=",vq,"\t",F.subs({p:vp,q:vq}))
```

# Ex 3 Implication Truth Table

```python
from sympy import Implies
from sympy.logic.boolalg import truth_table
```

```python
I=Implies(p,q)
```

```python
print(pretty(p),"\t",pretty(q),"\t",pretty(I))
for row in truth_table(I,[p,q]):
    vL,vr=row
    vp,vq=vL
    print(vp,"\t",vq,"\t",vr)
```

The only row where $p \Rightarrow q$ is *False* is where p is *True* and q is *False*, whitch means we canoot start with a *True* Statement and end up with a *False* statement, *True* Always Implies *True*.

# Ex 3 bis

```python
import pandas as pd
def Truth_Table(Exprs,Vars):
    A=[]
    for row in truth_table(Vars[0],Vars):
        vL,vr=row
        A.append(vL)
    for exp in Exprs:
        for i,row in enumerate(truth_table(exp,Vars)):
            _,vr=row
            A[i].append(vr)
    return pd.DataFrame(A,columns=Vars+[pretty(exp) for exp in Exprs])
Truth_Table([p|q,p&q,p>>q],[p,q])
```

# Ex 4 Biconditional (↔) Equivalent

```python
from sympy import Equivalent
```

```python
Eq=Equivalent(p,q)
print(pretty(p),"\t",pretty(q),"\t",pretty(Eq))
for row in truth_table(Eq,[p,q]):
    vL,vr=row
    vp,vq=vL
    print(vp,"\t",vq,"\t",vr)
```

# Ex 5 Operator Precedence

```python
r=symbols("r")
```

```python
A=(~p&q)|r
B=~p&(q|r)
print(pretty(A),"\t\t",pretty(B))
D={p:True,q:False,r:True}
print(A.subs(D),"\t\t\t",B.subs(D))
```

# Ex 7 Tautology Detection satisfiable

```python
from sympy import satisfiable
```

```python
def is_tautology(expr) -> bool:
    return not satisfiable(Not(expr))
```

```python
Formules7=[p|~p,(p>>q)>>(~q>>~p),p&~p,(p|q)>>p,(p&q)>>p]
D7={"Is Tautology":["✅" if  is_tautology(expr) else  "❌"  for expr in Formules7]}
pd.DataFrame(D7,index=[pretty(F) for F in Formules7])
```

# Ex 8 Contradiction Detection satisfiable

```python
def is_contradiction(expr) -> bool:
    return not satisfiable(expr)
```

```python
Formules7=[p&~p,p|~p,(p>>q)&p&~q]
D7={"Is Contradiction":["✅" if  is_contradiction(expr) else  "❌"  for expr in Formules7]}
pd.DataFrame(D7,index=[pretty(F) for F in Formules7])
```

# Ex 10 Logic Gates — NAND and NOR gates

```python
from sympy.logic import Nand,Nor
```

```python
Truth_Table([Nand(p,q),Nor(p,q)],[p,q])
```

```python
def NAND(a,b):
    return ~(a&b)
```

```python
def NOR(a,b):
    return ~(a|b)
```

```python
def NEG_byNand(p):
    return Nand(p,p)
```

```python
NEG_byNand(p)
```

```python
def AND_byNand(p,q):
    return simplify_logic(Nand(Nand(p,q),Nand(p,q)))
```

```python
AND_byNand(p,q)
```

```python
def OR_byNand(p,q):
    return NEG_byNand(AND_byNand(Nand(p,p),Nand(q,q)))
```

```python
OR_byNand(p,q)
```

```python
from sympy.logic import Equivalent
```

```python
is_tautology(Equivalent(~p,NEG_byNand(p)))
```

```python
is_tautology(Equivalent(p|q,OR(p,q)))
```

```python
is_tautology(Equivalent(p|q,OR(p,q)))
```

```python
is_tautology(Equivalent(p&q,AND_byNand(p,q)))
```

```python
def NEG_byNOR(p):
    return NOR(p,p)
```

```python
NEG_byNOR(p)
```

```python
def AND_byNOR(p,q):
    return NOR(NOR(p,p),NOR(q,q))
```

```python
AND_byNOR(p,q)
```

```python
def OR_byNOR(p,q):
    return NOR(NOR(p,q),NOR(p,q))
```

```python
OR_byNOR(p,q)
```

```python
is_tautology(Equivalent(p&q,AND_byNOR(p,q)))
```

```python
is_tautology(Equivalent(p|q,OR_byNOR(p,q)))
```

```python
is_tautology(Equivalent(~p,NEG_byNOR(p)))
```

Single-Gate Computers: Apollo Guidance Computer
Why engineers did this
1. Manufacturing Efficiency: Buying thousands of identical parts is cheaper than buying small quantities of many different parts.
2. Reliability: It is easier to rigorously test and perfect the production of one specific circuit design.
3. Simplicity: Standardizing the voltage and timing across the entire machine becomes much easier when every component behaves identically.
